## Objects: basics

A corpus is of class `cx.Corpus`. It contains a collection of texts, which are of class `cx.Text`:

```python
print(type(corpus))
print(corpus)
```

```python
text = corpus[84]   # alternatively: text = corpus.texts[84]
print(type(text))
print(text)
```

A `cx.Text` contains a list of sentences, which are of class `cx.Sentence`:

```python
sentence = text[17]   # alternatively: sentence = text.sentences[17]
print(type(sentence))
print(sentence)
```

Each `cx.Sentence` is made up of tokens (`cx.Token`). Each `cx.Token` may have zero or more analyses (`cx.Analysis`). Finally, analyses consist of morphemes (`cx.Morpheme`).

```python
token = sentence[3]   # alternatively: token = sentence.tokens[3]
print(type(token))
print(token)
```

```python
analysis = token[0]   # alternatively: analysis = token.ana[0]
print(type(analysis))
print(analysis)
```

```python
morpheme = analysis[2]   # alternatively: morpheme = analysis.morphemes[3]
print(type(morpheme))
print(morpheme)
```

All the “inner” items of an object (e.g. all tokens in a sentence) are stored in a `list` (except for `cx.Morpheme`). Therefore, it is possible to:
- iterate over an object with a loop
- get to a specific item using an index (alternatively, use the method `.get_id()`)
- get the number of items with `len()`

```python
print("Iterate over tokens of a sentence:")
for t in sentence:
    print(t.id, t)
```

```python
print("Get an item using the object’s id:")
print(sentence[3])
print(sentence.get_id(4))   # Note that ids start at 1, not 0!
```

```python
print("Get the number of items:")
print(len(sentence))
```

If a text or a corpus is correctly imported, “inner” items will contain information about what “outer” objects they belong to:

```python
print(morpheme)
print("->", morpheme.back_ana)
print("->", morpheme.back_token)
print("->", morpheme.back_sent)
print("->", morpheme.back_text)
```

## Object attributes

All objects have their own attributes. For instance, if you have a `cx.Morpheme`, you can easily extract its morph (phonological form), its gloss and its POS tag. The object also contains information on how it is written in the glossing in the form of “prefix” and “suffix” (i.e. punctuation that precedes and follows the morpheme itself).

```python
print(morpheme)

print("morph:", morpheme.morph)
print("gloss:", morpheme.gloss)
print("_prefix:", morpheme._prefix)
print("_suffix:", morpheme._suffix)
print("morph_full:", morpheme.morph_full)
print("gloss_full:", morpheme.gloss_full)
print("pos:", morpheme.pos)
```

An `cx.Analysis` can use these attributes of each morpheme to give a glossing-like representation of the morphological structure:

```python
print(analysis)
print("morph_string:", analysis.morph_string)
print("gloss_string:", analysis.gloss_string)
print("view():", analysis.view())
```

Similarly, `cx.Token`, `cx.Sentence`, `cx.Text` and `cx.Corpus` also have useful attributes (these are just a few examples):

```python
print(token)
print("token:", token.token)
print("wtype:", token.wtype)
```

```python
print(sentence)
print("text:", sentence.text)
print("translation:", sentence.translation)
print("comment:", sentence.comment)
print("meta:", sentence.meta)
print("timestamps:", sentence.timestamps)
```

```python
print(text)
print("title:", text.title)
print("filename:", text.filename)
print("filetype:", text.filetype)
print("meta:", text.meta)
print("audio:", text.audio)
```

## Search

### Search morpheme

To search for a morpheme with certain properties, use the method `.search_morpheme()`. This method can be called from a `cx.Corpus`, a `cx.Text`, a `cx.Sentence` or a `cx.Token`.

Provide either of these conditions as arguments: `morph` (phonological form), `gloss` (gloss or lemma), `pos` (POS tag). If several conditions are provided, all apply simultaneously. All three arguments may contain regular expressions, and `gloss` is case-sensitive.

```python
results = corpus.search_morpheme(morph="atā")
results
```

```python
results = corpus.search_morpheme(gloss="father")
results
```

```python
results = corpus.search_morpheme(morph="atā", gloss="TIME2")
results
```

```python
results = corpus.search_morpheme(gloss="CAUSE")
results
```

Search results are stored not in a simple list, but in a special object `cx.MorphemeSearch`. This object can help you browse the results in different ways. For example, you can get the corresponding analyses or sentences to gain more context:

```python
results.as_ana
```

```python
results.as_sent
```

Alternatively, you can browse them as glossed examples:

```python
print(*results.as_glossing, sep="\n\n")
```

You can get all unique morph-gloss combinations to get a generalized idea of the result:

```python
results.morphemes
```

Finally, it is possible to export the results into a `pandas` dataframe:

```python
results.to_df()
```

### Search morpheme chain

If you need to look for specific combinations of morphemes, use the method `.search_morphemechain()`. It will find all cases, where the required morphemes are consecutive in a token’s analysis.

Provide an argument `gloss_index_string`, which is a glossed string in the Tsakorpus format. It has all glosses and morphs for each morpheme, separated by a hyphen: `"go{sit}-INF{t}-PURP{ow}"`). All glosses and morphs may contain regular expressions, and glosses are case-sensitive. You don’t need to specify hyphens and equal signs for affixes and clitics.

```python
stroka = "PURP{ow}-DAT{ard}"
results = corpus.search_morphemechain(gloss_index_string=stroka)
results
```

```python
results = corpus.search_morphemechain(gloss_index_string=".+{carāng}-PRS..+{.+}")
results
```

Results of morpheme chain search are stored in an object of the type `cx.MorphemeChainSearch`, so you can view the results in the similar neat ways as `cx.MorphemeSearch` (see above):

```python
results.as_ana
```

```python
print(*results.as_glossing, sep="\n\n")
```

### Search token

You can search for individual tokens with the method `.search_token()`. This method can be called from a `cx.Corpus`, a `cx.Text` or a `cx.Sentence`.

Provide an argument `token` (the full phonological form of the token). It may contain regular expressions. The search will ignore case and punctuation marks on the sides of the token:

```python
results = corpus.search_token(token="uz=um")
results
```

If you want a search that is sensitive to case or punctuation, you can specify this in additional arguments:

```python
results = corpus.search_token(token="Uz=um…", ignore_case=False, ignore_punct=False)
results
```

Alternatively, if you want an even more flexible search, you can specify that punctuation marks that denote morpheme boundaries should also be ignored. In that case, you can use a query with no morpheme boundaries and get results:

```python
results = corpus.search_token(token="sutum", ignore_morphemes=True)
results
```

Results of token search are stored in an object of the type `cx.TokenSearch`, so you can view the results in the similar neat ways as `cx.MorphemeSearch` (see above):

```python
results.as_sent
```

You can also get all unique analyses to get a generalized idea of the result:

```python
results = corpus.search_token(token="čūɣ̌ǰat", ignore_morphemes=True)
results.analyses
```

### Search in a subcorpus

You can look for morphemes, morpheme chains or tokens in specific parts of the corpus. To do this, use the argument `text_filter` in a search to specify metadata values of texts. For each metadata category, you can specify one or several values that need to be included in the search:

```python
results = corpus.search_token(token="nok.+", text_filter={
    "subcorpus": "Pear Film retellings"
})
results.as_sent
```

```python
results = corpus.search_morpheme(gloss="brother.*", text_filter={
    "subcorpus": ["Oral texts (old recordings)", "Texts (Zarubin 1960)"]
})
results.as_ana
```

You can look up metadata values using a `cx.Corpus` method `.meta_values()`:

```python
corpus.meta_values(key="subcorpus")
```

For example, let us compare the frequency of the Russian word *što* ‘what’ in texts of different periods (unfortunately, only for texts for which the exact year of creation is known):

```python
corpus.meta_values(key="year")
```

```python
results = corpus.search_morpheme(morph="što", text_filter={
    "year": list(range(1910, 1960))
})
results.as_sent
```

```python
results = corpus.search_morpheme(morph="što", text_filter={
    "year": list(range(2010, 2026))
})
results.as_sent
```

## Export

### Export search results to `csv`

To export search results to a new `csv` file, first export them into a `pandas` dataframe and then call method `.to_csv()` from this dataframe:

```python
df = results.to_df()
df.to_csv("name_for_new_table.csv", index=False)
```

### Export corpus to `csv`

You can export the full corpus into a (quite large!) `csv` file. It will contain all texts, sentences, tokens and morphemes. To do it, call the method `.to_df()` from your `cx.Corpus` and then export the resulting dataframe into `csv`:

```python
full_df = corpus.to_df()
# full_df.to_csv("name_for_full_corpus_table.csv", index=False)
```

```python
full_df.head()
```

### Export subcorpus or individual text to `csv`

You can export a certain subcorpus easily if you create a dataframe with all texts (see above) and then use `pandas` to filter out certain values of metadata categories:

```python
legends_df = full_df[full_df.subcorpus == "Legends (Alamshoev 1994)"]
legends_df
```

Similarly, to export one text, you can call `.to_df()` from a `cx.Text` object.

```python
three_brothers_df = text.to_df()
three_brothers_df.head()
```

### Export text to other formats

You can export a `cx.Text` to other formats. Crucially, it is possible to export them to Tsakorpus-formatted `json` or to the standard `eaf` for ELAN:

```python
text.to_json()
text.to_eaf()
```

Method `.to_print()` will print the text sentence by sentence, with translations if there are any.

```python
print(text.to_print())
```

Method `.to_latex()` will export the text into glossed examples for the LaTeX library `expex`:

```python
print(text.to_latex())
```

These two methods can also be called from a `cx.Sentence`:

```python
print(sentence.to_print())
print(sentence.to_latex())
```
